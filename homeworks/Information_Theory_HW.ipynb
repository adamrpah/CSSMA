# Information Theory Homework

In the `data/text_data` folder you have the canonical 20 News groups dataset: `20ng-test-all-terms.txt`. On each line of this text file, the first word is the newsgroup that the post was made and what follows are the processed terms that were in the post. In your homework, your goal is replicate and implement the method in Gerlach, Shi, and Amaral *A universal information theoretic approach to the identification of stopwords*. This requires completing three exercises:

1. Calculate the conditional entropy of each term (each post will be as a document)
2. Implement the random null model
3. Calculate the Information content for each word.
