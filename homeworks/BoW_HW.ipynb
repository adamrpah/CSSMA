# Bag of Words Homework

In the data folder there is a text document that contains all of the works William Shakespeare: `data/text_data/Shakespeare.txt`.  For your homework, you should compare the word usage between a comedy ("All's Well that End's Well") and a drama ("The Tragedy of Macbeth"). To do this you will have to complete these three steps:

1. Extract and clean all dialogue from "All's Well that End's Well"
2. Extract and clean all dialogue from "The Tragedy of Macbeth"
3. Calculate and plot the [rank-frequency](https://en.wikipedia.org/wiki/Rank–size_distribution) plot for both of the plays on the same graph.
