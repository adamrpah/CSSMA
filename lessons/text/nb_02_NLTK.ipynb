# Stemming

Stemming is the removal of conjugations to get at the base form of a word (i.e., so `running` and `run` will be recognized as the same word). As is shown in the associated research paper readings, we are performing these changes algorithmically which means that there will be variations depending on the algorithm chosen. 

```python
from nltk import stem, tokenize

wordlist = tokenize.word_tokenize(fulltext[:500])


dir(stem)
```

```python
ls = stem.LancasterStemmer()

print([ls.stem(w) for w in wordlist])
```

```python
ps = stem.PorterStemmer()

print([ps.stem(w) for w in wordlist])
```

```python
ss = stem.SnowballStemmer('english')

print([ss.stem(w) for w in wordlist])
```

```python
import nltk
nltk.download('wordnet')

ws = stem.WordNetLemmatizer()

print([ws.lemmatize(w) for w in wordlist])
```

# Extending beyond a bag of words

So far we have only dealt with what are called `uni-grams` (i.e. single word, bag of words). Bigrams and trigrams are also a part of the picture and are a different tokenization approach. 

We didn't cover them at the start, but we can access them through the n-grams functions. 

```python
list(nltk.bigrams(fulltext[:500]) )
```

```python
list( nltk.bigrams(wordlist) )
```

Why do bigrams matter? Typically it is in the context of some other analysis or relationship (i.e. some statistical learning module as a feature vector). They handicap your ability to do a direct analysis (two items instead of one); however, they expand your ability to model structure.

There are a number of reasons and instances where you will want to rely on Xgrams instead of or in addition to the unigram bag of words approach in order to have a description of the text that takes structure into account. A common reason is when you have a multi-word concept that encodes meaning, which often happens in specialized fields/writing. 

Just because you perform the extraction as multiple words, doesn't mean that you cannot reduce the Xgram  to a single word token. For convenience in further processing these tokens will be joined as a single string like so:

```python
['-'.join(x) for x in list(nltk.bigrams(wordlist))]
```

You wouldn't join unigrams and bigrams in a bag-of-words analysis, but you would do so in applications where you are generating features from the text. 

# POS

The final basic preprocessing step is determining the parts of speech. This is the lead in to may other machine learning techniques that leverage parts of speech to determine structure and novelty

```python
nltk.pos_tag(nltk_regex.tokenize('The fox ran quickly to its prey'))
```

How does this work? The simple answer is that this is a pre-trained model built from an annotated corpus. Text from that corpus has been trained like so:

`[[('today','NN'),('is','VBZ'),('good','JJ'),('day','NN')], [('yes','NNS'),('it','PRP'),('beautiful','JJ')]]`

and the model takes that learning and makes predictions on newly submitted text. The NLTK pos_tagger is trained on The Wall Street Journal corpus (clear links to source here: https://stackoverflow.com/questions/32016545/how-does-nltk-pos-tag-work/41384824)

# Alternative Packages

There are of course more modern alternatives to NLTK that rely on language models (as opposed to classical algorithms). While NLTK has trained models in it, it's not the core foundation of the package like these newer packages. The big packages would be:

* spaCy
* huggingface

`huggingface` is really more of a package to make it easy to use and train LLMs, so it's not really a competitor to NLTK in a pure sense. `spaCy` is however a very opinionated library that mimics a lot of the core functionality in NLTK while being based on a language model.

Now we can use the package and the downloaded language model

```python
import spacy
nlp = spacy.load('en_core_web_sm')

fulltext = open('../../data/Othello.txt').read()
#We have spacy process the text into a document
doc = nlp(fulltext[:500])
doc
```

```python
#We can get out sentences
for sentence in doc.sents:
    print(sentence)
```

```python
#Or noun chunks
for nchunk in doc.noun_chunks:
    print(nchunk)
```

```python
#We can look at our tokens
for token in doc:
    print(token.text, token.pos_, token.dep_)
```

```python
#And our statistically predicted entities
for ent in doc.ents:
    print(ent, ent.label_)
```
